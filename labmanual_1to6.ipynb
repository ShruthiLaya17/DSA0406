QUESTION 1

```python
import numpy as np

student_scores = np.array([
    [80, 75, 85, 70],
    [90, 85, 80, 75],
    [70, 80, 75, 85],
    [85, 90, 88, 80]
])

average = np.mean(student_scores, axis=0)

print("Average marks:", average)

subjects = ["Math", "Science", "English", "History"]
highest = subjects[np.argmax(average)]

print("Subject with highest average:", highest)
```

QUESTION 3

```python
import numpy as np

house_data = np.array([
    [3, 1200, 250000],
    [5, 2000, 450000],
    [6, 2500, 550000],
    [4, 1800, 350000],
    [7, 3000, 650000]
])

houses = house_data[house_data[:, 0] > 4]

average_price = np.mean(houses[:, 2])

print("Average sale price:", average_price)
```

QUESTION 4

```python
import numpy as np

sales_data = np.array([10000, 12000, 15000, 18000])

total_sales = np.sum(sales_data)

percentage_increase = ((sales_data[3] - sales_data[0]) /
                       sales_data[0]) * 100

print("Total sales:", total_sales)
print("Percentage increase:", percentage_increase, "%")
```

QUESTION 5

```python
import numpy as np

fuel_efficiency = np.array([30, 35, 40, 45])

average = np.mean(fuel_efficiency)

model1 = fuel_efficiency[0]
model2 = fuel_efficiency[1]

improvement = ((model2 - model1) / model1) * 100

print("Average fuel efficiency:", average)
print("Percentage improvement:", improvement, "%")
```

QUESTION 6

```python
prices = [100, 200, 150]
quantities = [2, 1, 3]

discount_rate = 10
tax_rate = 5

subtotal = sum(p * q for p, q in zip(prices, quantities))

discount = subtotal * discount_rate / 100
after_discount = subtotal - discount

tax = after_discount * tax_rate / 100
total_cost = after_discount + tax

print("Subtotal:", subtotal)
print("Discount:", discount)
print("Tax:", tax)
print("Total cost:", total_cost)
```
