```python

import pandas as pd
import re
import matplotlib.pyplot as plt
from collections import Counter

data = pd.DataFrame({
    'feedback': [
        'The product is good and the service is good',
        'The service is fast and the product is useful',
        'Good product and good quality',
        'The service is excellent'
    ]
})

stop_words = {'the', 'and', 'is', 'a', 'to', 'of', 'in', 'it'}

text = ' '.join(data['feedback'].dropna()).lower()
words = re.findall(r'\b[a-z]+\b', text)
words = [word for word in words if word not in stop_words]

frequency = Counter(words)

n = int(input("Enter N: "))
top_words = frequency.most_common(n)

print("Top frequent words:")
for word, count in top_words:
    print(word, count)

if top_words:
    labels = [word for word, count in top_words]
    values = [count for word, count in top_words]

    plt.bar(labels, values)
    plt.xlabel('Words')
    plt.ylabel('Frequency')
    plt.title('Top N Frequent Words')
    plt.show()
```

QUESTION 19

```python


import numpy as np
from scipy import stats

drug = np.array([10, 12, 8, 11, 9, 13, 10, 12, 11, 9])
placebo = np.array([4, 5, 3, 6, 4, 2, 5, 3, 4, 6])

def confidence_interval(data):
    n = len(data)
    mean = np.mean(data)
    std = np.std(data, ddof=1)

    t_value = stats.t.ppf(0.975, n - 1)
    error = t_value * std / np.sqrt(n)

    lower = mean - error
    upper = mean + error

    return round(float(lower), 4), round(float(upper), 4)

print("Drug group 95% CI:", confidence_interval(drug))
print("Placebo group 95% CI:", confidence_interval(placebo))
```

QUESTION 21

```python
import numpy as np
from scipy import stats

data = [4.8, 5.1, 5.4, 5.0, 5.3, 5.2, 4.9, 5.5, 5.1, 5.2]

n = int(input("Enter sample size (maximum 10): "))
confidence = float(input("Enter confidence level (95 or 99): "))
precision = float(input("Enter desired precision: "))

if n < 2 or n > len(data):
    print("Enter a sample size between 2 and 10.")
else:
    sample = data[:n]
    mean = np.mean(sample)
    std = np.std(sample, ddof=1)

    level = confidence / 100
    t_value = stats.t.ppf((1 + level) / 2, n - 1)
    margin = t_value * std / np.sqrt(n)

    print("Point Estimate:", round(mean, 3))
    print("Confidence Interval:",
          round(mean - margin, 3), "to",
          round(mean + margin, 3))
    print("Margin of Error:", round(margin, 3))

    if margin <= precision:
        print("Desired precision achieved.")
    else:
        print("Desired precision not achieved.")
```

```python

import pandas as pd
from scipy import stats
import numpy as np

data = pd.DataFrame({
    "rating": [4.0, 4.5, 4.2, 4.1, 4.4, 3.9, 4.3, 4.0, 4.5, 4.1]
})

ratings = data["rating"].dropna()

mean = ratings.mean()
std = ratings.std()
n = len(ratings)

t = stats.t.ppf(0.975, n - 1)
margin = t * std / np.sqrt(n)

print("Average rating:", round(mean, 2))
print("95% Confidence Interval:",
      (round(mean - margin, 2), round(mean + margin, 2)))
```

```python
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

control = [3, 4, 5, 4, 3, 5, 4, 3]
treatment = [7, 8, 6, 9, 7, 8, 9, 7]

t, p = ttest_ind(control, treatment)

print("Control mean:", sum(control) / len(control))
print("Treatment mean:", sum(treatment) / len(treatment))
print("P-value:", round(p, 4))

if p < 0.05:
    print("Significant treatment effect")
else:
    print("No significant treatment effect")

plt.bar(["Control", "Treatment"],
        [sum(control) / len(control),
         sum(treatment) / len(treatment)])
plt.ylabel("Mean outcome")
plt.title("Treatment Comparison")
plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score

data = load_iris()

X = data.data
y = data.target

features = input("Enter feature names separated by commas: ").split(",")
target = input("Enter target variable name: ")

X = data.frame if hasattr(data, "frame") else X

if target != "target":
    print("Using target variable: target")

columns = [name.strip() for name in features]
X = data.data[:, [list(data.feature_names).index(c) for c in columns]]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)

model = DecisionTreeClassifier(random_state=1)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, pred), 2))
print("Precision:", round(precision_score(
    y_test, pred, average="weighted", zero_division=0), 2))
print("Recall:", round(recall_score(
    y_test, pred, average="weighted", zero_division=0), 2))
print("F1-score:", round(f1_score(
    y_test, pred, average="weighted", zero_division=0), 2))
```
